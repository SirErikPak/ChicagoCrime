#  Concept 1: Stationarity
A time series is stationary if its statistical properties do not change over time. Specifically, three conditions must hold:
1. Constant mean - the average doesn't drift up or down
2. Constant variance - the spread of values doesn't grow or shrink
3. Constant autocovariance - the relationship between values at different lags depends only on the lag, not on when you are in the series

---
# Why does stationarity matter so much?
Most classical models (ARIMA, SARIMA, VAR) assume stationarity. If your series isn't stationary, the model's parameter estimates become unreliable, your standard errors are wrong, and your forecasts will drift off into nonsense. This is called spurious regression - a real, dangerous trap.

---
**Weak (covariance) stationarity** - the practical standard. Requires:
* E[Xₜ] = μ (constant mean)
* Var(Xₜ) = σ² (constant variance)
* Cov(Xₜ, Xₜ₋ₖ) depends only on lag k, not on t

**Strict stationarity** - the full joint distribution is time-invariant. Much stronger, rarely required in practice.

* Deterministic trend - the trend follows a fixed, predictable formula like Xₜ = α + βt + εₜ. The upward drift is baked into the equation itself. You remove it by regressing on time and subtracting the fitted line (detrending).
* Stochastic trend - the series wanders randomly with no fixed path: Xₜ = Xₜ₋₁ + εₜ. Each step is a random shock that permanently shifts the level. This is called a unit root. You remove it by differencing - computing Yₜ = Xₜ - Xₜ₋₁.

# ACF (Autocorrelation Function) 
* Measures how strongly a time series correlates with its own past values. At each lag 𝑘, the ACF quantifies the relationship between the current observation and the value 𝑘 months earlier. This makes the ACF a fundamental diagnostic tool for understanding the internal structure of a time series, especially when assessing whether the process is stationary, trending, or driven by seasonal patterns.
* A key property of the ACF is that it captures both direct and indirect effects. If Month 1 influences Month 2, and Month 2 influences Month 3, then Month 1 will appear to influence Month 3 even if there is no direct causal link. This "shadow" effect is a natural consequence of autocorrelation propagating through time. It means that the ACF at higher lags often reflects the accumulated influence of earlier months, not just immediate relationships. This is why the ACF is so effective at detecting long‑memory behavior.
* When a time series contains a trend, the ACF reveals it through a distinctive pattern:
    * a slow, gradual decay across many lags. Instead of dropping sharply after lag 1, the autocorrelation remains high for an extended sequence of lags. This indicates that shocks persist and the series mean drifts over time, which is the hallmark of non‑stationarity. A slow, linear decay in the ACF is one of the most reliable visual signatures of a trending process.
* In your Arson example, the ACF displays exactly this slow‑decay pattern. The autocorrelations remain elevated across multiple lags, confirming that the series has long‑range dependence and a drifting mean. This aligns perfectly with the KPSS test results, which also identified Arson as non‑stationary. Together, the ACF and KPSS provide converging evidence that Arson should be routed to decomposition, where trend and seasonal components can be explicitly modeled rather than treated as noise.

#  PACF (Partial Autocorrelation Function)
* Isolates the direct relationship between a time series and its lagged values. Unlike the ACF, which captures both direct and indirect effects, the PACF removes the influence of intermediate lags. This makes it a powerful diagnostic for identifying the true order of the autoregressive structure in a series. Conceptually, while the ACF answers “How correlated is Month 1 with Month 3 overall?”, the PACF answers "How correlated is Month 1 with Month 3 after removing the effect of Month 2?"
* Because the PACF filters out indirect pathways, it provides a clearer view of the underlying dependency structure. A large spike at lag 1 followed by a sharp drop suggests that the series is primarily driven by a first‑order autoregressive process. In contrast, multiple significant PACF spikes indicate deeper, multi‑step dependencies. This distinction is crucial when determining whether a series has meaningful structure or is simply noisy.
* The PACF is also a valuable tool for distinguishing between trend, seasonality, and noise. In trending series, the ACF decays slowly, but the PACF typically shows a strong spike at lag 1 and then rapidly diminishes. This pattern reflects the fact that trends create persistent correlations, but the direct influence of earlier lags is concentrated in the first few steps. In seasonal series, the PACF often exhibits pronounced spikes at seasonal intervals (e.g., lag 12 for monthly data), revealing repeating patterns that STL decomposition can extract.
* When applied to your crime data, the PACF complements the ACF and KPSS results. For example, Arson shows a slow ACF decay-indicating trend, but its PACF likely displays a dominant lag‑1 spike, confirming that the trend is driving most of the structure. This reinforces your routing decision to use decomposition. Conversely, crime types with weak or erratic PACF patterns are better suited for robust or presence‑based handling, since they lack a stable autoregressive structure.
