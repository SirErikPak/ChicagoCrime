# Working with MultiIndex DataFrame

```python
import pandas as pd
import numpy as np

# seed
SEED = 1776
```

## Sample Data

```python
# Create raw data 
np.random.seed(SEED)
# np.tile: construct a new array by repeating the input list a specific number of times
data = {
    'sector': np.repeat(['001', '002', '002'], 8),           # 3 * 8 = 24
    'district_code': np.tile(['001', '002', '003'], 8),      # 3 * 8 = 24
    'crime_code_desc': np.tile(['Aggravated Battery', 'Homicide', 'Burglary'], 8), # 3 * 8 = 24
    'indexed': np.tile(['I', 'N'], 12),                       # 2 * 12 = 24
    'count': np.random.randint(50, 1000, 24)    
}
# convert to dataframe
df = pd.DataFrame(data)

# MultiLevel Index
df = df.set_index(['sector', 'district_code', 'indexed'])

# display
df.head()
```
