## Summary

This exploratory visualization analysis of Chicago crime patterns from 2001 to 2025 complements the formal compositional modeling in ChicagoCrimeDetection by examining crime through spatial, temporal, and categorical lenses. Indexed crimes show strong seasonal concentration, with most violent and property offenses peaking during Quarter 3 (July–September), particularly in July. Violent crimes such as homicide, aggravated assault, and battery are most common on Sundays during late-night hours, while property crimes like burglary and larceny peak on Fridays during daytime periods. Although most indexed crimes follow a summer-driven seasonal cycle, some offenses display distinct timing anomalies, including Motor Vehicle Theft and Robbery peaking in October and Criminal Sexual Assault peaking in January.

Non-Indexed crimes follow different behavioral dynamics and are less tied to seasonal variation. Instead, their timing reflects human routines, nightlife activity, institutional schedules, and enforcement practices. Financial and administrative crimes such as fraud and embezzlement align with business-hour activity, while gambling, liquor-law violations, prostitution, and weapons violations cluster during evening and nighttime periods. Drug-related offenses show spring increases, and Weapons Violations rise sharply following COVID, reinforcing broader post-pandemic volatility observed elsewhere in the analysis.

The Z-score trend analysis reveals long-term structural changes in Chicago crime composition over the last two decades. Traditional crimes such as robbery, theft, prostitution, vandalism, battery, and liquor-law violations show a persistent long-run decline, remaining below their historical averages throughout much of the last decade. In contrast, Fraud shows a strong upward pivot beginning around 2016, suggesting a shift from physical street crime to financially and digitally enabled offenses. Violent crime patterns display substantial volatility, including the 2016 homicide spike and the larger 2020–2022 homicide and weapons-violation surge during the pandemic era. Motor Vehicle Theft shows one of the sharpest reversals in the dataset, shifting from historically low levels in 2019 to a major spike by 2023. The year 2020 stands out as a major anomaly period, with dramatic declines in public order and theft-related crimes alongside simultaneous spikes in homicide and weapons offenses.

Spatially, crime composition is highly concentrated across specific police districts and neighborhoods. Commercial districts such as Central and Near North dominate financial and theft-related offenses, while Englewood, Gresham, and South Chicago show persistent concentration in violent crimes. Harrison emerges as both a violent-crime and drug-enforcement hotspot. At the neighborhood level, Englewood leads in 10 of 26 major offense categories, primarily violent and property crimes, while Austin leads in 9 categories concentrated in vice, drug, and quality-of-life offenses. This geographic concentration suggests that citywide compositional analyses may partially aggregate across very different local crime regimes.

Location-description analysis further highlights differences between crime types. Indexed crimes are overwhelmingly street-centered, particularly larceny-theft, motor vehicle theft, robbery, assault, and homicide. In contrast, Non-Indexed crimes are more evenly divided between public and domestic environments. Fraud, family-related offenses, and sexual-abuse-related crimes occur most frequently in residences, while drug violations, prostitution, vandalism, and weapons offenses remain concentrated in public street settings. Simple Battery uniquely peaks in apartment settings, indicating a stronger association with multi-unit residential environments.

Several interpretive cautions are important. Some categories, particularly Drug Abuse Violations, Weapons Violations, Gambling, Prostitution, and Liquor Laws, are heavily influenced by enforcement practices and officer-initiated activity rather than purely underlying criminal incidence. Additionally, ward-level analysis was excluded because of major upstream geographic assignment inconsistencies in the source data. Overall, the findings present a coherent picture of long-run crime evolution in Chicago: traditional street and property crimes have generally declined, financial and fraud-related offenses have increased, post-pandemic violent volatility has intensified, and crime composition remains strongly shaped by geography, human activity patterns, and enforcement behavior.

Specifically, Ward '50' contained 694,069 incidents ($\approx$2× the next-highest ward) whose latitudes spanned the full north-south extent of Chicago (41.64–42.02) rather than the actual Ward 50 geography (~42.00, far north side), indicating an upstream fallback where un-geocoded coordinates were defaulted to a single ward. Filtering Ward 50 would remove both the misassigned rows and ~91K legitimate West Ridge incidents, with no reliable way to distinguish between them. Rather than present ward-level findings derived from data on this issue, we omit the Ward analysis and rely on the Police District, Neighborhood, and Location Description sections, which use independent geographic keys and were verified to be consistent with spatial findings.

##### Helper Function

```python
def top_crime(data: pd.DataFrame, indexed: bool, top: int, 
              column: dict={"fbi_code_desc": "FBI Crime Type", "district_location": "Police District"}) -> None:

    # Get Keys
    it = iter(column)
    fbi_key = next(it, None)
    next_key  = next(it, None)

    data_out = (
        data.reset_index(name='count')
            .assign(**{fbi_key:  lambda d: d[fbi_key].astype(str),
                       next_key: lambda d: d[next_key].astype(str)})
            .sort_values([fbi_key, "count"], ascending=[True, False])
            .groupby(fbi_key).head(top).reset_index(drop=True)
            ).set_index([fbi_key, next_key]).sort_index(level=[0,1])
    
    # convert boolean argument
    flag = 'I' if indexed else 'N'
    txt  = 'Indexed' if indexed else 'Non-Indexed'
    bool_mask = data_out.indexed.eq(flag)

    # Rename Index
    data_out = data_out.rename_axis(index=column)
    # Convert to DataFrame and rename the column 
    data_out = data_out.rename(columns={"count": "Incident Count"})

    print(f"{'-'*25} Top {top} {txt} Crime {'-'*25}\n")
    print(data_out.loc[bool_mask].iloc[:, -1].apply(lambda x: f"{x:,}"))
```

#### Helper Functions

```python
def top_category(df, sort_col):
    return (
        df.sort_values(["fbi_code_desc", "count"], ascending=[True, False])
          .groupby("fbi_code_desc", as_index=False)
          .first()
          .sort_values(sort_col)
    )
```

```python
# Combine
out_tod   = top_category(melt_tod, "time_of_day").drop(columns="count")
out_week  = top_category(melt_week, "day_of_week").drop(columns="count")
out_month = top_category(melt_month, "month").drop(columns="count")
out_qtr   = top_category(melt_qtr, "quarter").drop(columns="count")

out = (
    out_tod
    .merge(out_week,  on="fbi_code_desc", how="left")
    .merge(out_month, on="fbi_code_desc", how="left")
    .merge(out_qtr,   on="fbi_code_desc", how="left")
)
mapping = {
    'Late Night': '12AM - 03AM',
    'Early Morning': '04AM - 07AM',
    'Morning': '08AM - 11AM',
    'Afternoon': '12PM - 03PM',
    'Evening': '04PM - 07PM',
    'Night': '08PM - 11PM'
}

# Add mapping
out['time_of_day'] = out['time_of_day'] + ': ' + out['time_of_day'].map(mapping)

# Build mapping of fbi_code_desc -> indexed (I/N)
map_df = (
    df_crime_filtered[['fbi_code_desc', 'indexed']]
      .drop_duplicates()
)

# Merge mapping into out
out = out.merge(map_df, on='fbi_code_desc', how='left')
```

#### Calculating the Z-score highlights specialized anomalies-places where a specific crime type is occurring at a rate far beyond what is statistically normal for the rest of the city.
- A Z-score of 3.0
    - The value is higher than about 99.87% of all values in a normal distribution 
    - It’s an extreme outlier
- A Z‑score of 2 means:
    - The value is higher than about 97.7% of all values in a normal distribution
    - That year’s count is significantly higher than the historical average
- A Z-score of -1:
    - The value is lower than about 84% of all values
    - It’s a mild decrease, not extreme
